1. What is nf-core?

nf-core is a community project that develops, curates and maintains bioinformatics pipelines written in Nextflow. All pipelines follow shared guidelines and the same template, are tested automatically, run in containers, and are documented in a consistent way. This makes them portable across a laptop, an HPC cluster and the cloud, and makes their results reproducible.

2. How many pipelines are there currently in nf-core?

3. Are there any non-bioinformatic pipelines in nf-core?

Yes
such as nf-core/rangeland, which processes remotely sensed satellite imagery for earth observation and land-cover trend analysis.

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A pipeline (workflow) is a defined series of computational steps in which the output of one step becomes the input of the next, for example QC → trimming → alignment → quantification. We use pipelines to:

automate multi-step analyses,
process many samples the same way,
run independent steps in parallel,
resume after failures,
make analyses reproducible and shareable.

5. Why do you think nf-core adheres to strict guidelines?

So that every pipeline meets the same standard of quality, reproducibility and usability, whoever wrote it.

6. What are the main features of nf-core pipelines?

Built with Nextflow: parallel execution, -resume, and runs on any compute environment.
Software in containers (Docker, Singularity) or conda, with pinned versions.
A common template and shared, reusable modules.
Parameter and samplesheet validation through a JSON schema.
A test profile with small test data, plus automated tests (CI) on GitHub.
Versioned releases and standard documentation (usage, output).

Check out the current directory. Next to the outdir you specified, what else has changed?

Nextflow created:

work/: one subfolder per task, named by a hash (e.g. work/3f/a2b1c9...), holding the task’s input links, outputs, script (.command.sh) and logs. This is what -resume uses.
.nextflow/: Nextflow’s cache database and history of runs.
.nextflow.log: the log of the latest run. Older logs are renamed .nextflow.log.1, .2, and so on.

results/pipeline_info/ also gained execution reports for each run.

```python
# delete the work directory and run the pipeline again using -resume. What did change?
!nextflow run nf-core/differentialabundance -profile test,docker -r 2.0.0 --outdir result3 -resume
```

What changed?

Nothing was cached this time, and every task ran again from the start, as in the first run. -resume relies on the task outputs stored in work/. The cache in .nextflow/ still knew about the earlier tasks, but their output files were gone, so they couldn’t be reused.

## Lets look at the results

### What is differential abundance analysis?

It tests, for each feature (e.g. gene, transcript or protein), whether its abundance differs significantly between groups of samples, such as treated vs control. For each feature it estimates the size of the change (log2 fold change) and its significance (a p-value, corrected for testing many features at once, e.g. FDR). For RNA-seq counts this is usually done with DESeq2 or limma. “Differential abundance” is the general term; “differential expression” is the RNA-seq case.

Give the most important plots from the report:

PCA plot: samples should group by condition, and replicates should lie close together. An outlier or a batch effect shows up here first.

Sample clustering / correlation heatmap: shows which samples are most similar.

Abundance distributions (boxplots/density plots before and after normalisation): check that normalisation made the samples comparable.
