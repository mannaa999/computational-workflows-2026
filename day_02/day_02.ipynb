```python
import pandas as pd
import openpyxl

# Load the conditions table
conditions = pd.read_excel("conditions_runs_oxy_project.xlsx")

# Load the number of bases for each run
base_counts = pd.read_csv("base_counts.csv")

# Add Bases to the conditions table using the Run column
conditions = conditions.merge(base_counts, on="Run", how="left")

# Sort by number of bases, then by condition
conditions = conditions.sort_values(
    by=["Bases", "condition: Sal", "Condition: Oxy"]
).reset_index(drop=True)

display(conditions)

# Display the final table
```
