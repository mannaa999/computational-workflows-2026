### 1. Describe the concept of Workflows, Processes and Channels we deal with in Nextflow

Workflows: A workflow is the top-level structure that defines how the pieces of an analysis connect. It declares which processes run and how data flows between them through channels, but it doesn't say when each step runs. Nextflow works that out itself from the data dependencies: a step runs as soon as its inputs are available, and independent steps run in parallel automatically.

Processes: A process is a single step of a workflow that runs one task, usually a command-line tool such as FastQC or salmon quant. Each process declares its input, its output, and a script block containing the shell command to run. Processes are isolated from each other: each task runs in its own working directory (inside work/), optionally in its own container, and processes can only exchange data through channels. 

Channels: Channels are the connections that carry data between processes. A channel is like a queue or conveyor belt: one end produces items (file paths, values, or tuples like a sample ID with its read files), and processes consume them as they arrive. Channels are created with factories such as Channel.of(...) or Channel.fromPath('*.fastq.gz'), and they can be transformed with operators like map, filter, or collect. There are two kinds: queue channels, which are consumed item by item and emptied as they are read, and value channels, which hold a single value that can be read any number of times (useful for a reference index used by every sample).

### Why was CONVERTTOUPPER run so often?

The split process emits several chunk files, and they enter the next process’s queue channel as separate items. Nextflow launches one task per item, so CONVERTTOUPPER ran once for each chunk, in parallel. This is how Nextflow parallelises work: split the data, process each piece independently, then collect the results.
